```python
numero1 = int(input("inserta el número 1: ")) #guarda en una variable lo que pide el usuario con un número entero
numero2 = int(input("inserta el número 2: "))

def sum(numero1, numero2): #Aqui lo que hace es sumar las dos variables que se guardaron y se pidieron
    result = numero1 + numero2
    print(numero1, "+", numero2, "=", result)

def sub(numero1, numero2): #Aqui lo que hace es restar las dos variales que se guardaron y se pidieron
    result = numero1 - numero2
    print(numero1, "-", numero2, "=", result)

def mult(numero1, numero2): #Aqui lo que hace es multiplicar las dos variables que se guardaron y se pidieron
    result = numero1 * numero2
    print(numero1, "*", numero2, "=", result)

def div(numero1, numero2): #Aqui lo que se hace es dividir las dos variables que s guardaron y se pidieron
    if (numero2 != 0): #El if lo que hace es de que si es un 0 no se podra dividir con otro numero o con un 0
      result = numero1 / numero2
      print(numero1, "/", numero2, "=", result)
    else:
      print("el numero 2 no es igual a 0 y no se puede dividir por 0") #Y la funcion del else lo que hace es devolver un mensaje que diga que no se pude dividir entre 0

sum(numero1,numero2)
sub(numero1,numero2)
mult(numero1,numero2)
div(numero1,numero2)
#Esta parte ejecuta el resultado de la variables sum, sub, mult y div o en pocas palabras finalizar la funcion de las variables nombradas
```
